```python
import pandas as pd
```

```python
def gen_merged_games_df(year, season):
    
    """
    Accepts year and season (Summer or Winter)
    Merges the athlete_events and noc_regions CSV files on NOC into a Pandas dataframe
    Resulting dataframe includes valid countries and filtered on year and season
    """
    
    athlete_events = pd.read_csv('athlete_events.csv')
    noc_regions = pd.read_csv('noc_regions.csv')
    df = athlete_events.merge(noc_regions, on='NOC', how='inner')
    condition = str(year) + ' ' + season
    games_df = df[df.Games==condition]
    
    return games_df
```

```python
def gen_noc_medal_table(df):
    
    """
    Accepts Pandas dataframe
    Generate non-null unique values of region, Event, and Medal in a dataframe
    Add Gold, Silver, and Bronze flag columns - indicates medal color if it matches, null if not
    Add count columns for overall number of medals as well as Gold, Silver, and Bronze medals
    Return dataframe with unique counts and sort by Gold Medal Count and Silver Medal Count
    """
    
    noc_df = df.loc[:, ['region', 'Event', 'Medal']].dropna().drop_duplicates()
    
    noc_df['Gold'] = noc_df['Medal'].map({'Gold': 'Gold'})
    noc_df['Silver'] = noc_df['Medal'].map({'Silver': 'Silver'})
    noc_df['Bronze'] = noc_df['Medal'].map({'Bronze': 'Bronze'})
    noc_df['Medal_Count'] = noc_df.groupby(['region'])['Medal'].transform('count')
    noc_df['Gold_Medal_Count'] = noc_df.groupby(['region'])['Gold'].transform('count')
    noc_df['Silver_Medal_Count'] = noc_df.groupby(['region'])['Silver'].transform('count')
    noc_df['Bronze_Medal_Count'] = noc_df.groupby(['region'])['Bronze'].transform('count')
    
    noc_medal_table = noc_df.loc[:, ['region', 'Medal_Count', 'Gold_Medal_Count', 'Silver_Medal_Count', 'Bronze_Medal_Count']].drop_duplicates().sort_values(by=['Gold_Medal_Count','Silver_Medal_Count'], ascending=False)
    
    return noc_medal_table
```
