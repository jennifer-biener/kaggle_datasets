```python
#Number of lines each character has
data.loc[:, ['Character', 'Line']].groupby(['Character']).count()
```

```python
data['Character'] = data['Character'].str.replace('\"', '')
data['Character'].unique()
```

```python
data.loc[:, ['Character', 'Line']].groupby(['Character']).count().sort_values(by='Line', ascending=False).head(10)
```

```python
data['Character'] = data['Character'].str.replace('Michel', 'Michael')
data['Character'] = data['Character'].str.replace('Micheal', 'Michael')
data['Character'] = data['Character'].str.replace('Micahel', 'Michael')
data['Character'] = data['Character'].str.replace('Micael', 'Michael')
data['Character'] = data['Character'].str.replace('Michal', 'Michael')
data['Character'].unique()
```

```python
data['Character'] = data['Character'].str.strip()
```

```python
data.loc[:, ['Character', 'Line']].groupby(['Character']).count().sort_values(by='Line', ascending=False).head(10)
```

```python
data['Character'].unique()
```

```python
data['Character'] = data['Character'].str.replace('.', '')
data['Character'] = data['Character'].str.replace('Dwightkschrute', 'Dwight')
data['Character'].unique()
```

```python
data.loc[:, ['Character', 'Line']].groupby(['Character']).count().sort_values(by='Line', ascending=False).head(10)
```
