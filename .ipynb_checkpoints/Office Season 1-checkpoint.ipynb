```python
import pandas as pd
```

```python
df = pd.read_csv('the-office_lines.csv')
df.head()
```

```python
season_1_df = df[df['Season'] == 1]
season_1_df.info()
```

```python
season_1_df['Character'].unique()
```

```python
season_1_df[(season_1_df.Character == 'Michael') | (season_1_df.Character == 'Michel')]
```

```python
season_1_df[season_1_df.Character == 'Michel']
```

```python
season_1_df['Character'] = season_1_df['Character'].str.replace('Michel', 'Michael')
season_1_df['Character'].unique()
#season_1_df['Character'] = season_1_df['Character'].str.replace('Todd Packer', 'Packer')
```

```python
season_1_df[(season_1_df.Character == 'Todd Packer') | (season_1_df.Character == 'Packer')]
```

```python
season_1_df = season_1_df.rename(columns={"Unnamed: 0":"Line_Number"})
season_1_df.head()
```

```python
season_1_df[(season_1_df.Line_Number == 1595) |(season_1_df.Line_Number == 1596) |(season_1_df.Line_Number == 1597) | (season_1_df.Line_Number == 1598) | (season_1_df.Line_Number == 1599)]
```

```python
season_1_df[(season_1_df.Episode_Number == 5) & (season_1_df.Line_Number > 1580) & (season_1_df.Line_Number < 1620)]
```

```python
#season_1_df['Character'] = season_1_df['Character'].str.replace('Packer', 'Todd Packer')
season_1_df['Character'] = season_1_df['Character'].str.replace('Todd Todd Packer', 'Todd Packer')
season_1_df['Character'].unique()
```

```python
season_1_df[season_1_df['Character'] == 'Michael And Dwight']
```

```python
season_1_line_cnt = season_1_df.loc[:, ['Character', 'Line_Number']].groupby(['Character']).count().sort_values(by='Line_Number', ascending=False)
season_1_line_cnt
```

```python
ax = season_1_line_cnt.plot.bar(y='Line_Number')
```

```python
season_1_ep_unique = season_1_df.loc[:, ['Character', 'Episode_Number']].drop_duplicates()
season_1_ep_unique
```

```python
season_1_ep_cnt = season_1_ep_unique.groupby(['Character']).count().sort_values(by='Episode_Number', ascending=False)
season_1_ep_cnt
```

```python
season_1_char_per_episode = season_1_ep_unique.groupby(['Episode_Number']).count().sort_values(by='Episode_Number', ascending=True)
season_1_char_per_episode
```

```python
ax2 = season_1_ep_cnt.plot.bar(y='Episode_Number')
ax3 = season_1_char_per_episode.plot.bar(y='Character')
```

```python
season_1_lines_per_ep = season_1_df.loc[:, ['Line_Number', 'Episode_Number']].groupby(['Episode_Number']).count()
season_1_lines_per_ep
```

```python
ax4 = season_1_lines_per_ep.plot.bar(y='Line_Number')
```
