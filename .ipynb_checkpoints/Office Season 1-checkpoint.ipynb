```python
import pandas as pd
```

```python
df = pd.read_csv('the-office_lines.csv')
df.head()
```

```python
season_1_df = df[df['Season'] == 1]
season_1_df.info()
```

```python
for i in season_1_df['Unnamed: 0'].values:
    print(i)
#There is a null value between 1971 and 1972

season_1_df['Unnamed: 0'].count
```

```python
for i in season_1_df['Season'].unique():
    print(i)
```

```python
for i in season_1_df['Episode_Number'].unique():
    print(i)
```

```python
for i in season_1_df['Character'].unique():
    print(i)
    
"""
Checking for off values:
Michel
Michael And Dwight
Packer
"""
```

```python
season_1_df[season_1_df.Character == 'Michel']
```

```python
season_1_df['Character'] = season_1_df['Character'].str.replace('Michel', 'Michael')
season_1_df[season_1_df.Character == 'Michel']
#season_1_df.head()
```

```python
season_1_df[season_1_df.Character == 'Michael And Dwight']
```

```python
season_1_df[season_1_df.Character == 'Packer']
```

```python
season_1_df['Character'] = season_1_df['Character'].str.replace('Todd Packer', 'Packer')
#Replacing Packer with Todd Packer resulted in one Todd Todd Packer value
season_1_df[season_1_df.Character == 'Todd Packer']
#season_1_df.head()
```

```python
season_1_df = season_1_df.rename(columns={"Unnamed: 0":"Line_Number"})
season_1_df.head()
```

```python
season_1_line_cnt = season_1_df.loc[:, ['Character', 'Line_Number']].groupby(['Character']).count().sort_values(by='Line_Number', ascending=False)
season_1_line_cnt
```

```python
season_1_line_cnt['Line_Number'].sum()
```

```python
#Plot bar of Characters vs the Number of Lines in Season 1
ax = season_1_line_cnt.plot.bar(y='Line_Number')
```

```python
pie = season_1_line_cnt.plot.pie(y='Line_Number', figsize=(11, 6))
```

```python
#season_1_df.loc[:, ['Character', 'Episode_Number']]
season_1_ep_unique = season_1_df.loc[:, ['Character', 'Episode_Number']].drop_duplicates()
season_1_ep_unique
```

```python
#season_1_ep_unique.groupby(['Character']).count().sort_values(by='Episode_Number', ascending=False)
season_1_ep_cnt = season_1_ep_unique.groupby(['Character']).count().sort_values(by='Episode_Number', ascending=True)
season_1_ep_cnt
```

```python
ax2 = season_1_ep_cnt.plot.barh(y='Episode_Number')
```

```python
season_1_char_per_episode = season_1_ep_unique.groupby(['Episode_Number']).count()#.sort_values(by='Episode_Number', ascending=True)
season_1_char_per_episode
```

```python
ax3 = season_1_char_per_episode.plot.bar(y='Character')
#pie3 = season_1_char_per_episode.plot.pie(y='Character') #Pie chart won't work here b/c it deals with a sum instead of a percentage
```

```python
season_1_lines_per_ep = season_1_df.loc[:, ['Line_Number', 'Episode_Number']].groupby(['Episode_Number']).count()
season_1_lines_per_ep
```

```python
ax4 = season_1_lines_per_ep.plot.bar(y='Line_Number')
pie4 = season_1_lines_per_ep.plot.pie(y='Line_Number')
```
