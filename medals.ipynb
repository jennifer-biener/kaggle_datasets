```python
import pandas as pd
```

```python
df = pd.read_csv('athlete_events.csv')
df.head()
```

```python
#medals by games
df.groupby('Games')['Medal'].count()
```

```python
#plot medals by games
games_medals = df.loc[:, ['Games', 'Medal']].groupby(['Games']).count()
games_medals.plot.bar()
```

```python
summer = df[df.Season=='Summer']
winter = df[df.Season=='Winter']
```

```python
summer_medals = summer.loc[:, ['Games', 'Medal']].groupby(['Games']).count()
summer_medals.plot.bar()
```

```python
winter_medals = winter.loc[:, ['Games', 'Medal']].groupby(['Games']).count()
winter_medals.plot.bar()
```

```python
sports_medals = df.loc[:, ['Sport', 'Medal']].groupby(['Sport']).count()
sports_medals.sort_values(by='Medal', ascending=False).head(10)
```

```python
sports_medals.plot.bar()
```

```python
events_medals = df.loc[:, ['Event', 'Medal']].groupby(['Event']).count()
events_medals.sort_values(by='Medal', ascending=False).head(10)
```

```python
noc_medals = df.loc[:, ['NOC', 'Medal']].groupby(['NOC']).count()
noc_medals.sort_values(by='Medal', ascending=False).head(10)
```

```python
noc_medals.plot.pie(y='Medal')
```
