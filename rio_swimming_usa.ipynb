```python
import pandas as pd
```

```python
athlete_events = pd.read_csv('athlete_events.csv')
noc_regions = pd.read_csv('noc_regions.csv')
df = athlete_events.merge(noc_regions, on='NOC', how='inner')
df.head()
```

```python
rio_swimming_usa = df[(df.NOC=='USA') & (df.Games=='2016 Summer') & (df.Sport=='Swimming')]
rio_swimming_usa
```

```python
rio_swimming_usa['Event'].value_counts()
#number of athletes in each race
```

```python
rio_swimming_usa_results = rio_swimming_usa.loc[:, ['Event', 'Medal']].dropna().drop_duplicates()
#Needed to determine if medal was earned in an event
rio_swimming_usa_results
```
