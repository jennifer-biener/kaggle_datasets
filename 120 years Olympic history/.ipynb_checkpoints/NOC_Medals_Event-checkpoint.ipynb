```python
import pandas as pd
```

```python
athlete_events = pd.read_csv('athlete_events.csv')
athlete_events.head()
```

```python
noc_regions = pd.read_csv('noc_regions.csv')
noc_regions.head()
```

```python
df = athlete_events.merge(noc_regions, on='NOC', how='inner')
df.head()
```

```python
df.count()
```

```python
df.loc[:, ['NOC', 'Games', 'Event', 'Medal']]
```

```python
df.loc[:, ['NOC', 'Games', 'Event', 'Medal']][df.Games=='2016 Summer']
```

```python
usa = df[df.NOC=='USA']
```

```python
usa.head()
```

```python
#medal is by athlete, I want to find medal by event
usa['Event'].unique()
```

```python
usa['Event'].value_counts() #athletes in each event
```

```python
usa_2016 = usa[usa.Games=='2016 Summer']
usa_2016.head()
```

```python
usa_2016['Event'].value_counts()
```

```python
usa_2016['Athlete_Counts'] = usa_2016.groupby(['Event'])['ID'].transform('count')
usa_2016.head()
```

```python
usa_2016.loc[:, ['Event', 'Medal']]
```

```python
usa_2016.loc[:, ['Event', 'ID']].groupby(['Event']).count()
```

```python
usa_2016_swimming = usa_2016[usa_2016.Sport=='Swimming']
```
