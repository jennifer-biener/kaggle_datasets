```python
"""
Compute medal tables for each games. 
The general viewership is usually interested in seeing where their country places each Game
"""
```

```python
import pandas as pd
from ipynb.fs.full.functions import *
```

```python
df = gen_merged_df()
df.head()
```

```python
#years_list = [year for year in df[df.Season=='Summer']['Year'].sort_values().unique()]
#years_list
```

```python
#Summer Games Medal Tables
years_list = [year for year in df[df.Season=='Summer']['Year'].sort_values().unique()]

for year in years_list:
    games_df = gen_game_df(df, year, 'Summer')
    noc_medal_table = gen_noc_medal_table(games_df)
    ax = noc_medal_table.head(20).plot.bar(x='region', y=['Gold_Medal_Count', 'Silver_Medal_Count', 'Bronze_Medal_Count'], stacked=True, figsize=(11,5), title=str(year)+' Summer Games Medal Count')
```

```python
#Winter Games Medal Tables
years_list = [year for year in df[df.Season=='Winter']['Year'].sort_values().unique()]

for year in years_list:
    games_df = gen_game_df(df, year, season='Winter')
    noc_medal_table = gen_noc_medal_table(games_df)
    ax = noc_medal_table.head(20).plot.bar(x='region', y=['Gold_Medal_Count', 'Silver_Medal_Count', 'Bronze_Medal_Count'], stacked=True, figsize=(11,5), title=str(year)+' Winter Games Medal Count')
"""    
1998 Olympics - Russia and Norway are switched, with Norway missing a gold medal
2014 Olympics - Latvia not shown; dataset generated before gold medal reallocation
"""
```
