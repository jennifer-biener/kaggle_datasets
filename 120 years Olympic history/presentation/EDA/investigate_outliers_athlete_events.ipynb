```python
"""
Objectives for athlete_events data:

- Distributions of numerical and categorical columns using seaborn histograms, kde
- Outliers of numerical columns by plotting boxplot on top of histograms
"""
```

```python
import pandas as pd
import seaborn as sns
from matplotlib import rcParams
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../athlete_events.csv')
df.head()
```

```python
df.describe(include='all')
```

```python
sns.set_theme()
rcParams['figure.figsize'] = 11,5
sns.histplot(data=df, x='Age', kde=True)
```

```python
sns.histplot(data=df, x='Age', binwidth=5, kde=True)
```

```python
sns.histplot(data=df, x='Height', kde=True)
```

```python
sns.histplot(data=df, x='Height', binwidth=5, kde=True)
```

```python
sns.histplot(data=df, x='Weight', kde=True)
```

```python
sns.histplot(data=df, x='Weight', binwidth=5, kde=True)
```

```python
sns.distplot(df['Weight'], kde=True, kde_kws={"color": "g", "alpha": 0.3, "linewidth": 5, "shade": True})
```

```python
sns.distplot(df['Height'], kde=True, kde_kws={"color": "r", "alpha": 0.3, "linewidth": 5, "shade": True})
```

```python
sns.distplot(df['Age'], kde=True, kde_kws={"color": "b", "alpha": 0.3, "linewidth": 5, "shade": True})
```

```python
f, (ax_box, ax_hist) = plt.subplots(2, sharex=True, gridspec_kw={"height_ratios": (.15, .85)})
sns.boxplot(df["Age"], ax=ax_box)
sns.histplot(data=df, x="Age", binwidth=5, kde=True, ax=ax_hist)
```

```python
f, (ax_box, ax_hist) = plt.subplots(2, sharex=True, gridspec_kw={"height_ratios": (.15, .85)})
sns.boxplot(df["Height"], ax=ax_box)
sns.histplot(data=df, x="Height", binwidth=5, kde=True, ax=ax_hist)
```

```python
f, (ax_box, ax_hist) = plt.subplots(2, sharex=True, gridspec_kw={"height_ratios": (.15, .85)})
sns.boxplot(df["Weight"], ax=ax_box)
sns.histplot(data=df, x="Weight", binwidth=5, kde=True, ax=ax_hist)
```
