```python
"""
Objectives for athlete_events data:

- Distribution of weight using seaborn histograms, kde
- Find out how distributions differ based on factors such as gender, sport, medalists, games, event
"""
```

```python
import pandas as pd
import seaborn as sns
from matplotlib import rcParams
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../athlete_events.csv')
df.head()
```

```python
df['Weight'].describe()
```

```python
df['Weight'].unique()
```

```python
sns.set_theme()
rcParams['figure.figsize'] = 11,5
```

```python
#Overall Height distribution - all athletes
f, (ax_box, ax_hist) = plt.subplots(2, sharex=True, gridspec_kw={"height_ratios": (.15, .85)})
sns.boxplot(df["Weight"], ax=ax_box)
sns.histplot(data=df, x="Weight", binwidth=5, kde=True, ax=ax_hist)
```

```python
#Weight distribution by gender - all athletes
sns.histplot(data=df[df.Sex=='F'], x="Weight", binwidth=5, color="skyblue", label="Female Weight", kde=True)
sns.histplot(data=df[df.Sex=='M'], x="Weight", binwidth=5, color="red", label="Male Weight", kde=True)

plt.legend() 
plt.show()
```

```python
#Weight distribution by gender - medalists
sns.histplot(data=df[(df.Sex=='F') & df.Medal.notnull()], x="Weight", binwidth=5, color="skyblue", label="Female Weight", kde=True)
sns.histplot(data=df[(df.Sex=='M') & df.Medal.notnull()], x="Weight", binwidth=5, color="red", label="Male Weight", kde=True)

plt.legend() 
plt.show()
```

```python
#Weight distribution by gender - gold medalists
sns.histplot(data=df[(df.Sex=='F') & (df.Medal=='Gold')], x="Weight", binwidth=5, color="skyblue", label="Female Weight", kde=True)
sns.histplot(data=df[(df.Sex=='M') & (df.Medal=='Gold')], x="Weight", binwidth=5, color="red", label="Male Weight", kde=True)

plt.legend() 
plt.show()
```

```python
#Weight distribution by gender, sport
for sport in df['Sport'].sort_values().unique():
    
    sns.histplot(data=df[(df.Sex=='F') & (df.Sport==sport)], x="Weight", binwidth=5, color="skyblue", label="Female Weight: "+sport, kde=True)
    sns.histplot(data=df[(df.Sex=='M') & (df.Sport==sport)], x="Weight", binwidth=5, color="red", label="Male Weight: "+sport, kde=True)

    plt.legend() 
    plt.show()
```
