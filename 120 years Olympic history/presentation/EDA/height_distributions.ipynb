```python
"""
Objectives for athlete_events data:

- Distribution of height using seaborn histograms, kde
- Find out how distributions differ based on factors such as gender, sport, medalists, games, event
"""
```

```python
import pandas as pd
import seaborn as sns
from matplotlib import rcParams
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../athlete_events.csv')
df.head()
```

```python
df['Height'].describe()
```

```python
df['Height'].unique()
```

```python
sns.set_theme()
rcParams['figure.figsize'] = 11,5
```

```python
#Overall Height distribution - all athletes
f, (ax_box, ax_hist) = plt.subplots(2, sharex=True, gridspec_kw={"height_ratios": (.15, .85)})
sns.boxplot(df["Height"], ax=ax_box)
sns.histplot(data=df, x="Height", binwidth=5, kde=True, ax=ax_hist)
```

```python
#Height distribution by gender - all athletes
sns.histplot(data=df[df.Sex=='F'], x="Height", binwidth=5, color="skyblue", label="Female Height", kde=True)
sns.histplot(data=df[df.Sex=='M'], x="Height", binwidth=5, color="red", label="Male Height", kde=True)

plt.legend() 
plt.show()
```

```python
#Height distribution by gender - medalists
sns.histplot(data=df[(df.Sex=='F') & df.Medal.notnull()], x="Height", binwidth=5, color="skyblue", label="Female Height", kde=True)
sns.histplot(data=df[(df.Sex=='M') & df.Medal.notnull()], x="Height", binwidth=5, color="red", label="Male Height", kde=True)

plt.legend() 
plt.show()
```

```python
#Height distribution by gender - gold medalists
sns.histplot(data=df[(df.Sex=='F') & (df.Medal=='Gold')], x="Height", binwidth=5, color="skyblue", label="Female Height", kde=True)
sns.histplot(data=df[(df.Sex=='M') & (df.Medal=='Gold')], x="Height", binwidth=5, color="red", label="Male Height", kde=True)

plt.legend() 
plt.show()
```

```python
df['Sport'].sort_values().unique()
```

```python
#Height distribution by gender, sport
for sport in df['Sport'].sort_values().unique():
    
    sns.histplot(data=df[(df.Sex=='F') & (df.Sport==sport)], x="Height", binwidth=5, color="skyblue", label="Female Height: "+sport, kde=True)
    sns.histplot(data=df[(df.Sex=='M') & (df.Sport==sport)], x="Height", binwidth=5, color="red", label="Male Height: "+sport, kde=True)

    plt.legend() 
    plt.show()
```

```python
#Height distribution by event in Athletics. No need to put M, F on same plot since events are different
for event in df[df.Sport=='Athletics']['Event'].sort_values().unique():
    
    sns.histplot(data=df[(df.Sport=='Athletics') & (df.Event==event)], x="Height", color="red", label=event, kde=True)

    plt.legend() 
    plt.show()
```
