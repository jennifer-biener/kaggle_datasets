```python
#Import required modules
import pandas as pd
import numpy as np
```

```python
"""
Create athlete_events dataframe and view sample data output.

ID represents a unique athlete entry and is a primary key.
"""
df = pd.read_csv('../../athlete_events.csv')
df.head()
```

```python
#Generate descriptive statistics on numeric columns in the dataframe
#df.describe(include='all')
df.describe()
```

```python
"""
Most useful numeric statistics are Age, Height, and Weight.

Age (years):
- mean=25.5
- stdev=6.4
- min=10
- 25th percentile=21
- 50th percentile=24
- 75th percentile=28
- max=97

Height (cm):
- mean=175.3
- stdev=10.5
- min=127
- 25th percentile=168
- 50th percentile=175
- 75th percentile=183
- max=226

Weight (kg):
- mean=70.7
- stdev=14.3
- min=25
- 25th percentile=60
- 50th percentile=70
- 75th percentile=79
- max=214
"""
```

```python
#Plot boxplots of age, height, weight independent of gender
ax = df.boxplot(column=['Age', 'Height', 'Weight'], figsize=(11,5), grid=False, vert=False)
```

```python
#Plot boxplots of age, height, weight - female
female = df[df.Sex=='F'].boxplot(column=['Age', 'Height', 'Weight'], figsize=(11,5), grid=False, vert=False)
```

```python
#Plot boxplots of age, height, weight - male
male = df[df.Sex=='M'].boxplot(column=['Age', 'Height', 'Weight'], figsize=(11,5), grid=False, vert=False)
```
