```python
import pandas as pd
```

```python
df = pd.read_csv('../events_medals.csv')
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
for year in df['event_year'].unique():
    print(df[df.event_year==year])
```

```python
united_states = df[df.country=='United States']
united_states
```

```python
united_states.sort_values(by='event_year').plot.bar(x='event_year', y=['gold', 'silver', 'bronze'], stacked=True, figsize=(11,5))
```

```python
for c in df['country'].unique():
    df[df.country==c].sort_values(by='event_year').plot.bar(x='event_year', y=['gold', 'silver', 'bronze'], stacked=True, figsize=(11,5), title=c)
```
