```python
season_2_df['Character'] = season_2_df['Character'].str.replace('Todd Packer', 'Packer')
#Replacing Packer with Todd Packer resulted in one Todd Todd Packer value
season_2_df[season_2_df.Character == 'Todd Packer']
#season_2_df.head()
```

```python
season_2_df = season_2_df.rename(columns={"Unnamed: 0":"Line_Number"})
season_2_df.head()
```

```python
season_2_line_cnt = season_2_df.loc[:, ['Character', 'Line_Number']].groupby(['Character']).count().sort_values(by='Line_Number', ascending=False)
season_2_line_cnt
```

```python
season_2_line_cnt['Line_Number'].sum()
```

```python
#Plot bar of Characters vs the Number of Lines in Season 1
ax = season_2_line_cnt.plot.bar(y='Line_Number')
```

```python
pie = season_2_line_cnt.plot.pie(y='Line_Number', figsize=(11, 6))
```

```python
#season_2_df.loc[:, ['Character', 'Episode_Number']]
season_2_ep_unique = season_2_df.loc[:, ['Character', 'Episode_Number']].drop_duplicates()
season_2_ep_unique
```

```python
#season_2_ep_unique.groupby(['Character']).count().sort_values(by='Episode_Number', ascending=False)
season_2_ep_cnt = season_2_ep_unique.groupby(['Character']).count().sort_values(by='Episode_Number', ascending=True)
season_2_ep_cnt
```

```python
ax2 = season_2_ep_cnt.plot.barh(y='Episode_Number')
```

```python
season_2_char_per_episode = season_2_ep_unique.groupby(['Episode_Number']).count()#.sort_values(by='Episode_Number', ascending=True)
season_2_char_per_episode
```

```python
ax3 = season_2_char_per_episode.plot.bar(y='Character')
#pie3 = season_2_char_per_episode.plot.pie(y='Character') #Pie chart won't work here b/c it deals with a sum instead of a percentage
```
